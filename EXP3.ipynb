```python
import pandas as pd
import numpy as np
data = {
    'First name' : ['Aryan', 'Rohan', 'Riya', 'Yash', 'Siddini'],
    'Last name' : ['Singh', 'Agrawal', 'shah', 'Batia', 'Khan'],
    'Type' : ['Full-Time Employee', 'intern', 'Full Time Employee', 'Full Time Employee', 'Part Time Employee'],
    'Department' : ['Administration', 'Technical', 'Adiministration', 'technical', 'technical'],
    'yoe' :[2, 3, 5, 7, 6],
    'salary' : [20000, 5000, 10000, 20000, 15000]
}
df = pd.DataFrame(data)
print(df)
```

```python
table = pd.pivot_table(df, index=['Type', 'Department'],values='salary', aggfunc=['mean'])
print(table)
```

```python
table1 = pd.pivot_table(df, index = ['Type'],values='salary',aggfunc = ['sum','mean'])
print(table1)
```

```python
table2 = pd.pivot_table(df, index = ['Type'],values='salary',aggfunc = ['count'])
print(table2)
```

```python
table3 = pd.pivot_table(df, index = ['Type'],values='salary',aggfunc = ['std'])
print(table3)
```
