```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
cars_df=pd.read_csv('/content/mtcars-selected-columns.csv')
print(cars_df)
```

```python
a=plt.scatter(cars_df['mpg'],cars_df['wt'],marker='*')
plt.xlabel=('mpg')
plt.ylabel=('weight')
plt.title('Relationship between the weight of the car and mpg')
plt.show()
```

```python
trans = cars_df['am'].value_counts()
print(trans)
trans.plot.bar()
plt.xlabel=("Transmission")
plt.title=("Frequency distribution of transmission type")
plt.show()
```

```python
cars_df['mpg'].plot(kind='box')
plt.xlabel=('mpg')
plt.title('whisker plot to mpg')
plt.show()
```

```python
minvalue=np.min(cars_df['mpg'])
q1=np.percentile(cars_df['mpg'],25)
median=np.median(cars_df['mpg'])
q3=np.percentile(cars_df['mpg'],75)
maxvalue=np.max(cars_df['mpg'])
print("The minimum value=", minvalue)
print("Q1=",q1)
print("The median=",median)
print("Q3=",q3)
print("The maximum value=", maxvalue)
```
