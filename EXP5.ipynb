```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm

cars_df = pd.read_csv('/content/Cars93.csv')
print(cars_df)

fig, ax = plt.subplots(figsize=(10, 4))
ax.scatter(cars_df['Horsepower'], cars_df['Price'])
ax.set_xlabel('Horsepower')
ax.set_ylabel('Price')
plt.title("Scatter plot")
plt.show()

cars_df['Horsepower'].plot(kind='box')
plt.show()

q1 = cars_df["Horsepower"].quantile(0.25)
q3 = cars_df["Horsepower"].quantile(0.75)
iqr = q3 - q1
upper_bound = q3 + (1.5 * iqr)
lower_bound = q1 - (1.5 * iqr)

horsepower_arr = cars_df["Horsepower"]
outliers = horsepower_arr[(horsepower_arr < lower_bound) | (horsepower_arr >= upper_bound)]
print("The following are the outliers in the box plot horsepower \n", outliers)

cars_df['Price'].plot(kind='hist')
plt.show()

upper_index = cars_df[cars_df['Horsepower'] > upper_bound].index
cars_df.drop(upper_index, inplace=True)

lower_index = cars_df[cars_df['Horsepower'] < lower_bound].index
cars_df.drop(lower_index, inplace=True)

cars_df.info()

horsepower_arr = cars_df["Horsepower"]
lower_cap = np.percentile(horsepower_arr, 1)
upper_cap = np.percentile(horsepower_arr, 99)

outliers = horsepower_arr[(horsepower_arr > upper_cap) | (horsepower_arr < lower_cap)]
print("The following are the outliers in the boxplot horsepower: \n", outliers)

for i in cars_df['Horsepower']:
    if i < lower_bound:
        cars_df['Horsepower'] = cars_df['Horsepower'].replace(i, lower_cap)
    elif i > upper_bound:
        cars_df['Horsepower'] = cars_df['Horsepower'].replace(i, upper_cap)

print(cars_df.info())

m = np.mean(cars_df['Price'])
print("mean:", m)
for i in cars_df['Price']:
    if i < lower_bound or i > upper_bound:
        cars_df['Price'] = cars_df['Price'].replace(i, m)

m = cars_df['Price'].median()
print("median:", m)
for i in cars_df['Price']:
    if i < lower_bound or i > upper_bound:
        cars_df['Price'] = cars_df['Price'].replace(i, m)
```
