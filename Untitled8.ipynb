```python
%%writefile app.py
import fastapi
from fastapi import FastAPI
from pydantic import BaseModel
import joblib
import pandas as pd
import tensorflow as tf

app = FastAPI()

# Load assets (Ensure you have run your training script first to generate these files)
try:
    model = tf.keras.models.load_model('flight_delay_model.keras')
    preprocessor = joblib.load('preprocessor.joblib')
except Exception as e:
    model = None
    preprocessor = None

class FlightData(BaseModel):
    Month: int
    DayOfWeek: int
    DepTimeHour: int
    Distance: int
    Carrier: str
    Origin: str
    Dest: str

@app.get("/")
def home():
    return {"status": "online"}

@app.post("/predict")
def predict_delay(flight: FlightData):
    if model is None or preprocessor is None:
        return {"status": "Error", "message": "Model files not found. Please train and save the model first."}

    input_data = pd.DataFrame([flight.model_dump()])
    transformed_data = preprocessor.transform(input_data)
    prediction_prob = float(model.predict(transformed_data, verbose=0))

    is_delayed = prediction_prob > 0.5
    status = "Delayed" if is_delayed else "On Time"

    return {
        "status": status,
        "delay_probability": round(prediction_prob, 4),
        "confidence_percentage": round(prediction_prob * 100, 2) if is_delayed else round((1 - prediction_prob) * 100, 2)
    }
```

```python
%%writefile ui.py
import streamlit as st
import requests

st.set_page_config(page_title="Flight Delay Predictor")
st.title("🛫 Real-Time Flight Delay Classifier")

col1, col2 = st.columns(2)
with col1:
    month = st.slider("Month", 1, 12, 1)
    day_of_week = st.slider("Day of the Week", 1, 7, 1)
    dep_hour = st.slider("Scheduled Departure Hour", 0, 23, 12)
    distance = st.number_input("Flight Distance (miles)", min_value=100, max_value=5000, value=750)

with col2:
    carrier = st.selectbox("Airline Carrier", ["AA", "DL", "UA", "WN", "B6"])
    origin = st.selectbox("Origin Airport", ["JFK", "LAX", "ORD", "ATL", "DFW"])
    dest = st.selectbox("Destination Airport", ["MIA", "SFO", "SEA", "DEN", "BOS"])

if st.button("Analyze Delay Risk", type="primary"):
    payload = {
        "Month": month, "DayOfWeek": day_of_week, "DepTimeHour": dep_hour,
        "Distance": distance, "Carrier": carrier, "Origin": origin, "Dest": dest
    }
    try:
        # Note: In Colab, both apps run on the same instance, so UI can talk directly to localhost:8000 internally
        response = requests.post("http://127.0.0", json=payload)
        result = response.json()

        st.subheader("Prediction Results")
        if result.get("status") == "Delayed":
            st.error(f"🚨 **Delayed** ({result['confidence_percentage']}% confidence)")
        else:
            st.success(f"✅ **On Time** ({result['confidence_percentage']}% confidence)")
    except Exception as e:
        st.error(f"Could not connect to backend: {e}")
```
