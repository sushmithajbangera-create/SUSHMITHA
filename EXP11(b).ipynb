```python
# lambda()

max = lambda a, b: a if (a > b) else b
print(max(4, 5))
x = lambda a: a + 1
print(x(5))
x = lambda a, b, c: a + b + c
print(x(5, 6, 2))
```

```python
# map()

def add_4(x):
    return x + 4
list1 = [4, 6, 7, 8, 9]
list2 = list(map(add_4, list1))
print(list2)
```

```python
# filter()

def start_with_A(s):
    return s[0] == "A"
fruit = ["Apple", "Banana", "Pear", "Apricot", "Orange"]
filter_object = filter(start_with_A, fruit)
print(list(filter_object))
```

```python
# reduce()

from functools import reduce
def sum_vals(x, y):
    return x + y
list1 = [6, 7, 8, 9]
s = reduce(sum_vals, list1)
print(s)
```
