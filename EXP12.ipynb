```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import warnings

warnings.filterwarnings('ignore')
df = pd.read_csv('/content/Iris.csv')
df.head()
df.shape
df = df.drop(columns=['Id'], errors='ignore')
df.head()
df.describe()
df.info()
df['Species'].value_counts()
df.isnull().sum()
df['SepalWidthCm'].hist()
df['SepalLengthCm'].hist()

from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['Species'] = le.fit_transform(df['Species'])
correct = df.corr()
fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(correct, annot=True, ax=ax, cmap='coolwarm')

px.scatter(df, x='Species', y='PetalWidthCm')
px.line(df, x='Species', y='PetalWidthCm')
px.scatter(df, x='Species', y='PetalLengthCm')
px.scatter(df, x='Species', y='SepalLengthCm')
px.scatter(df, x='Species', y='SepalWidthCm')
px.scatter_matrix(df, color='Species', title='iris', dimensions=['SepalLengthCm', 'SepalWidthCm','PetalLengthCm', 'PetalWidthCm'])

colors = ['red', 'orange', 'blue']
species_labels = ['Iris-setosa', 'Iris-versicolor', 'Iris-virginica']

for i in range(3):
    x = df[df['Species'] == i]
    plt.scatter(x['SepalLengthCm'], x['SepalWidthCm'],c=colors[i], label=species_labels[i])
plt.xlabel('sepal length')
plt.ylabel('sepal width')
plt.legend()

from sklearn.model_selection import train_test_split
x = df.drop(columns=['Species'])
y = df['Species']
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size=0.30)

from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(xtrain, ytrain)
print('accuracy:', model.score(xtest, ytest) * 100)

from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(xtrain, ytrain)
print('accuracy:', model.score(xtest, ytest) * 100)
```
